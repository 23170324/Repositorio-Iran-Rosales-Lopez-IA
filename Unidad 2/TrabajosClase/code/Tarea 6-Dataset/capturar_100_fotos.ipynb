```python
import cv2
from pathlib import Path
import time
from PIL import Image
from IPython.display import display, clear_output

TOTAL_FOTOS = 100
INTERVALO_SEGUNDOS = 1.0
CUENTA_REGRESIVA = 3
CARPETA_SALIDA = Path(
    r'C:\Users\PC\OneDrive\Documentos\Repositorio-Iran-Rosales-Lopez-IA\Unidad 2\TrabajosClase\code\Tarea 6-Dataset'
) / 'fotos_originales'
CARPETA_SALIDA.mkdir(parents=True, exist_ok=True)
print(f'Las fotograf?as se guardar?n en: {CARPETA_SALIDA}')
```

```python
camara = cv2.VideoCapture(0, cv2.CAP_DSHOW)
if not camara.isOpened():
    camara.release()
    camara = cv2.VideoCapture(1, cv2.CAP_DSHOW)
if not camara.isOpened():
    raise RuntimeError('No se pudo abrir la c?mara. Cierra otras aplicaciones que la est?n usando.')

for restante in range(CUENTA_REGRESIVA, 0, -1):
    ok, frame = camara.read()
    if ok:
        clear_output(wait=True)
        print(f'La captura comienza en {restante} segundo(s)...')
        display(Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)))
    time.sleep(1)

capturas = 0
prox_foto = time.monotonic()
try:
    while capturas < TOTAL_FOTOS:
        ok, frame = camara.read()
        if not ok:
            print('No se pudo leer un cuadro de la c?mara.')
            break
        ahora = time.monotonic()
        if ahora >= prox_foto:
            capturas += 1
            ruta = CARPETA_SALIDA / f'rostro_{capturas:03d}.jpg'
            cv2.imwrite(str(ruta), frame, [cv2.IMWRITE_JPEG_QUALITY, 95])
            prox_foto += INTERVALO_SEGUNDOS
            estado = f'Foto {capturas} de {TOTAL_FOTOS} tomada. Pr?xima captura: {INTERVALO_SEGUNDOS:.0f} s.'
        else:
            segundos = max(0, prox_foto - ahora)
            estado = f'Preparando foto {capturas + 1} de {TOTAL_FOTOS}: {segundos:.1f} s restantes.'
        clear_output(wait=True)
        print(estado)
        display(Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)))
        time.sleep(0.05)
finally:
    camara.release()
    clear_output(wait=True)
    print(f'Captura terminada: {capturas} fotograf?as guardadas en {CARPETA_SALIDA}')
```
