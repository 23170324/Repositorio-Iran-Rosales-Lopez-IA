```python
import numpy as np

# 1. La Función de Activación (Función Escalón)
def funcion_escalon(suma_ponderada):
    return 1 if suma_ponderada >= 0 else 0

# 2. Datos de entrenamiento (Compuerta OR)
# Entradas: [x1, x2]
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Salidas esperadas
y_esperado = np.array([0, 1, 1, 1])

# 3. Inicializar parámetros
pesos = np.random.rand(2) # Dos pesos inicializados al azar
sesgo = np.random.rand(1)[0] # Un sesgo al azar
tasa_aprendizaje = 0.1 # Qué tan grandes son los pasos al ajustar (learning rate)
epocas = 10 # Cuántas veces repasará los datos

print("Iniciando entrenamiento del Perceptrón...\n")

# 4. Ciclo de Entrenamiento
for epoca in range(epocas):
    errores_totales = 0
    print(f"--- Época {epoca + 1} ---")
    
    for i in range(len(X)):
        # a) Multiplicar entradas por pesos y sumar el sesgo (producto punto)
        suma = np.dot(X[i], pesos) + sesgo
        
        # b) Pasar por la función de activación
        prediccion = funcion_escalon(suma)
        
        # c) Calcular el error (Lo que debió ser menos lo que fue)
        error = y_esperado[i] - prediccion
        
        if error != 0:
            errores_totales += 1
            # d) Regla de actualización: Ajustar pesos y sesgo
            # Es un concepto similar al descenso de gradiente para minimizar la función de costo
            pesos += tasa_aprendizaje * error * X[i]
            sesgo += tasa_aprendizaje * error
            
    print(f"Errores cometidos: {errores_totales}")
    
    # Si ya no comete errores, la línea recta ya separó bien los datos
    if errores_totales == 0:
        print("¡El perceptrón ha aprendido exitosamente!")
        break

# 5. Probar el modelo final
print("\n--- Resultados Finales ---")
for i in range(len(X)):
    suma_final = np.dot(X[i], pesos) + sesgo
    prediccion_final = funcion_escalon(suma_final)
    print(f"Entrada: {X[i]} -> Predicción: {prediccion_final} (Esperado: {y_esperado[i]})")
```
